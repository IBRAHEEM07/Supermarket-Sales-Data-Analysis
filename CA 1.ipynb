## Pre Processing.

### Step 1.

```python
df.drop(["Description"],axis= 1, inplace= True) ##Droped the Description Column 
```

```python
df.columns
```

### Step 2.

```python
df["Customer ID"] = df["Customer ID"].fillna("Guest") ### Filling the NAN values
```

```python
df.isnull().sum()
```

```python
df.shape
```

### Step 3. 

```python
df= df[~df["Invoice"].astype(str).str.startswith("C")] ## Revmoved the cancled invoices.
```

### Step 4

```python
df= df[(df["Quantity"]>0)& (df["Price"]>0)] ## Removed invalid values in Quantity and Prize column
```

```python
df["Quantity"].unique()[:100]
```

### Step 5

```python
df.drop_duplicates()
```

```python
df["InvoiceDate"]=pd.to_datetime(df["InvoiceDate"])
```

```python
df.info()
```

```python
df['Revenue'] = df['Quantity'] * df['Price'] ## added a revenue column    
```

```python
df['MonthName'] = df['InvoiceDate'].dt.month_name()
```

```python
df.head()
```

```python
# Remove extreme quantity outliers
Q1 = df['Quantity'].quantile(0.25)
Q3 = df['Quantity'].quantile(0.75)
IQR = Q3 - Q1

df = df[(df['Quantity'] >= Q1 - 1.5*IQR) & 
        (df['Quantity'] <= Q3 + 1.5*IQR)]
```

```python
df.shape
```

## Visualization

#### INSIGHT 1: Revenue Distribution by Country

```python
country_revenue = df.groupby('Country')['Revenue'].sum().sort_values(ascending=False)
country_revenue.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(10,6))
country_revenue.head(10).plot(kind='bar',color= "green")
plt.title("Top 10 Countries by Revenue")
plt.ylabel("Total Revenue")
plt.xlabel("Country")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### INSIGHT 2: Time-Based Sales Patterns

```python
monthly_revenue = df.groupby("MonthName")["Revenue"].mean().sort_values(ascending=False)
monthly_revenue
```

```python
plt.figure(figsize=(10,6))
monthly_revenue.plot(kind='line',color="red")
plt.title("Monthly Revenue Trend")
plt.ylabel("Revenue")
plt.xlabel("Month")
plt.show()
```

### INSIGHT 3: Price vs Quantity Relationship

```python
correlation = df[['Price', 'Quantity', 'Revenue']].corr()
print(correlation)
```

```python
plt.figure(figsize=(10,8))
sns.heatmap(df.corr(numeric_only=True), 
            annot=True, 
            cmap='coolwarm', 
            fmt=".2f")

plt.title("Full Numerical Feature Correlation Heatmap")
plt.show()
```

```python
import seaborn as sns

plt.figure(figsize=(8,6))
sns.regplot(x='Price', y='Quantity', data=df, scatter_kws={'alpha':0.3})
plt.title("Price vs Quantity Relationship")
plt.show()
```

### INSIGHT 4: Customer Behaviour Analysis.

```python
customer_summary = df.groupby('Customer ID').agg({
    'Revenue': 'sum',
    'Invoice': 'nunique',
    'Quantity': 'sum'
}).reset_index()

customer_summary.rename(columns={
    'Invoice': 'Total_Orders',
    'Quantity': 'Total_Items'
}, inplace=True)

customer_summary = customer_summary.sort_values(by='Revenue', ascending=False)
```

```python
plt.figure(figsize=(10,6))
customer_summary.head(10).plot(
    x='Customer ID', 
    y='Revenue', 
    kind='bar'
)
plt.title("Top 10 Customers by Revenue")
plt.ylabel("Revenue")
plt.show()
```

### Query 1 — Total Revenue by Country

```python
import pandas as pd

query = """
SELECT 
    Country,
    SUM(Quantity * Price) AS Total_Revenue
FROM cleaned_supermarket_data
GROUP BY Country
ORDER BY Total_Revenue DESC
"""

result = pd.read_sql(query, engine)

result
```

### Query 2 — Top 10 Best Selling Products

```python
query1 = """
SELECT 
    StockCode,
    SUM(Quantity) AS Total_Units_Sold
FROM cleaned_supermarket_data
GROUP BY StockCode
ORDER BY Total_Units_Sold DESC
LIMIT 10;
"""

top_products = pd.read_sql(query1, conn)
top_products
```

```python
# Monthly Revenue Analysis
query2 = """
SELECT 
    MonthName,
    SUM(Revenue) AS Monthly_Revenue
FROM cleaned_supermarket_data
GROUP BY MonthName
ORDER BY FIELD(MonthName,
'January','February','March','April','May','June',
'July','August','September','October','November','December');
"""

monthly_revenue = pd.read_sql(query2, conn)
monthly_revenue
```

```python
#Top Customers by Spending
query3 = """
SELECT 
    `Customer ID`,
    SUM(Revenue) AS Total_Spent
FROM cleaned_supermarket_data
GROUP BY `Customer ID`
ORDER BY Total_Spent DESC
LIMIT 10;
"""

top_customers = pd.read_sql(query3, conn)
top_customers
```
